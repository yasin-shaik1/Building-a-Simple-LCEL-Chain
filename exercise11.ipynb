```python
# ============================================================
# EXPERIMENT: LCEL CHAIN
# Prompt Template → LLM → Output Parser
# Using Groq + LangChain
# ============================================================

# ------------------------------------------------------------
# STEP 1: Install required packages
# ------------------------------------------------------------

!pip install -q langchain langchain-core langchain-groq


# ------------------------------------------------------------
# STEP 2: Import libraries
# ------------------------------------------------------------

import os
from getpass import getpass

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq


# ------------------------------------------------------------
# STEP 3: Load Groq API key
# ------------------------------------------------------------

groq_api_key = getpass("Enter your Groq API key: ")

if not groq_api_key:
    raise ValueError("❌ GROQ_API_KEY not provided.")

os.environ["GROQ_API_KEY"] = groq_api_key

print("✅ Groq API key loaded successfully.")


# ------------------------------------------------------------
# STEP 4: Create the LLM
# ------------------------------------------------------------

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0.7,
    max_tokens=150
)

print("✅ Groq LLM initialized.")


# ------------------------------------------------------------
# STEP 5: Create Prompt Template
# ------------------------------------------------------------

prompt = ChatPromptTemplate.from_template(
    "Write a very short story (2-3 sentences) about {topic}."
)


# ------------------------------------------------------------
# STEP 6: Create Output Parser
# ------------------------------------------------------------

parser = StrOutputParser()


# ------------------------------------------------------------
# STEP 7: Build LCEL Chain
# ------------------------------------------------------------

chain = prompt | llm | parser

print("✅ LCEL chain created.")


# ------------------------------------------------------------
# STEP 8: Define Topics
# ------------------------------------------------------------

topics = [
    "a dragon and a knight",
    "a lost spaceship"
]


# ------------------------------------------------------------
# STEP 9: Invoke Chain
# ------------------------------------------------------------

for topic in topics:

    print("\n" + "=" * 60)
    print("Topic:", topic)
    print("=" * 60)

    try:

        result = chain.invoke({
            "topic": topic
        })

        print("\nStory:")
        print(result)

    except Exception as e:

        print("❌ Error:", e)


# ------------------------------------------------------------
# STEP 10: Explanation
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("LCEL CHAIN STRUCTURE")
print("=" * 60)

print("""
Prompt Template
      ↓
    LLM
      ↓
StrOutputParser
      ↓
   String Output

LCEL expression:

chain = prompt | llm | parser
""")

print("✅ Experiment completed successfully.")
```
